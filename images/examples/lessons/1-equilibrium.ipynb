# Lesson 1: Mass action and equilibrium

**You will:** simulate a reversible binding reaction, watch it settle, and check that at equilibrium the forward and reverse rates are equal.

**The model** ([`equilibrium.bngl`](equilibrium.bngl)): `A + B <-> C`. By *mass action*, the forward reaction runs at rate `kf·A·B` (it needs an A to meet a B) and the reverse at `kr·C`. Start with 100 A, 50 B and no C.

Run each cell with **Shift+Enter**.

```python
import math

import matplotlib.pyplot as plt

import bngsim
```

```python
model = bngsim.Model.load("equilibrium.bngl")
result = bngsim.Simulator(model, method="ode").run(t_span=(0, 300), n_points=301)
t, A, B, C = result.time, result.observables["A"], result.observables["B"], result.observables["C"]

plt.plot(t, A, label="A")
plt.plot(t, B, label="B")
plt.plot(t, C, label="C")
plt.xlabel("time (s)")
plt.ylabel("molecules")
plt.legend()
plt.show()
```

## Why it stops changing

The amounts level off, but the reactions do not stop: A and B keep binding and C keeps falling apart. The amounts stop changing when the two rates are equal. Plot both rates:

```python
kf, kr = 0.001, 0.01
plt.plot(t, kf * A * B, label="forward rate, kf·A·B")
plt.plot(t, kr * C, label="reverse rate, kr·C")
plt.xlabel("time (s)")
plt.ylabel("reactions per second")
plt.legend()
plt.show()
```

## The exact answer

At equilibrium `kf (A0 - C)(B0 - C) = kr C`: a quadratic in C, which we can solve by hand and compare with the simulation. The ratio `C / (A·B)` is the equilibrium constant, `K = kf / kr`, whatever amounts we start from.

```python
A0, B0 = 100, 50
b = kf * (A0 + B0) + kr
C_exact = (b - math.sqrt(b * b - 4 * kf * kf * A0 * B0)) / (2 * kf)
print(f"C at equilibrium: exact {C_exact:.2f}, simulated {C[-1]:.2f}")
print(f"K = kf/kr: exact {kf / kr:.4f}, simulated {C[-1] / (A[-1] * B[-1]):.4f}")
```
